apply and map method

```python
def replace_gender_values(value):
    if value == "Male":
        return "M"
    elif value == "male":
        return "M"
    elif value == "Female":
        return "F"
    elif value == "female":
        return "F"
    elif value == "M":
        return "M"
    elif value == "F":
        return "F"

# data1["Gender"] = data1["Gender"].apply(replace_gender_values)
data1["Gender"] = data1["Gender"].map(replace_gender_values)
```

##### Merging and Concatenation

```python
df1 = pd.DataFrame({
    "id": [1,2,3,4,5,6,7,8],
    "Fruit": ["Mango", "Banana", "Apple", "Pineapple", "Strawberry","Sugarcane","Dragonfruit","Blueberry"],
    "Total": [10, 60, 7, 4, 100, 12, 40, 120]
})

df1
```

```python
df2 = pd.DataFrame({
    "id":[1,2,3,4,5,10,11,12],
    "Color": ["Yellow", "Yellow", "Red", "Brown", "Red","Green","Brown","Orange"],
    "Season": ["Summer", "All", "Winter", "Summer", "Winter", "Winter", "Summer", "All"]
})

df2
```

```python
merged_data = pd.merge(df1, df2, on="id", how="outer")
merged_data
```

Concatenation

```python
df3 = pd.DataFrame({
    "id":[20,21],
    "Fruit":["Leechie","Coconut"],
    "Total":[8,4],
    "Color":["Pink", "Brown"],
    "Season":["Summer", "All"]
})
df3
```

```python
concatenated_data = pd.concat([merged_data, df3], axis=0)
concatenated_data.reset_index(inplace=True)
concatenated_data
```
