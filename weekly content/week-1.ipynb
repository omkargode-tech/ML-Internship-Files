### Casting
Conversion of one data type variable into another data type

```python
# below v is float type variable
v = 10.87126876
print(v)
print(type(v),"\n")

# here we are converting variable v into int data type
v = int(v)
print(v)
print(type(v))
```

```python
q = "109.123"
print(type(q))
q = float(q)
print(q)
print(type(q))
```

### String 
String is bascially array or sequence of characters in python.


```python
# String index starts from 0 and can have negative index aswell, -1 represents last element

# I n d l a --> 5
# 0 1 2 3 4 --> (0-4)  indexing
# -5 -4 -3 -2 -1       negative indexing

s = "Indala"
print(s[::-1])
```

```python
# Slicing

s = "Indala College of engineering"

# Accessing First Element -
print(s[0])


# Accessing Last Element -
print(s[len(s)-1])


# slicing 
print(s[0:7])


# Reversing String
print(s[::-1])

```

```python
# String Concatenation - adding two strings using "+" operator

t = "Indala"
y = "College"

p = t + " " + y
print(p)
```

```python
# String Methods 

s = "indala College of engineering"

# makes all characters to upper case
print(s.upper())

# makes all characters to lower case
print(s.lower())

# makes First character to upper case
print(s.capitalize())

# returns count of character or string which is passed as argument
print(s.count("e"))

# s.split()
```

```python
ww = 15
dd = 2

print(f"Addition Operator\n{ww} + {dd} = {ww + dd}\n")

print(f"Subtraction Operator\n{ww} - {dd} = {ww - dd}\n")

print(f"Multiplication Operator\n{ww} * {dd} = {ww * dd}\n")

print(f"Division Operator\n{ww} / {dd} = {ww / dd}\n")

print(f"Modulo Operator\n{ww} % {dd} = {ww % dd}\n")

print(f"Exponentiation Operator\n{ww} ** {dd} = {ww ** dd}\n")

print(f"Floor Division Operator\n{ww} // {dd} = {ww // dd}\n")
```

### Control Statement
#### If-Else
It is a decision making statement that executes different blocks of code based on whether condition is True or False

```python
# if (condition):
#    statements - print, variable assignment
# elif (condition):
#    statements
# else:
#    statements
```

```python
a = 3
b = 2

if a < b:
    print(f"{a} is greater than {b}")
    print("if block")
elif a == b:
    print(f"{a} is equal to {b}")
    print("elif block")
else:
    print(f"{b} is greater than {a}")
    print("else block")

```

```python
num = int(input("Enter any number: "))

if num % 2 == 0:
    print(f"{num} is even number")
else:
    print(f"{num} is odd number")
```

### Loops
A loop is a statement that repeats a block of code multiple times until a codition is met


```python
# while loop
# It executes set of instruction (block) as long as condition is true

# variable  initialization ->  i = 0
# set condition -> while  condition is  true keep executing block
# updation ->  we update variable which we are comparing
```

```python
num = 2
i = 1
while i <= 10:
    print(f"{num} X {i} = {num * i}")
    i += 1
    # i = i + 1
```

```python
# For loop
# Used to iterate (go through) over a sequence or iterable

s = "Indala"

for i in range(0,len(s)):
    print(s[i])


print("\n")
j = 0
while j < len(s):
    print(s[j])
    j += 1
```

### break continue pass

```python
# break is used to get out of the loop - breaking loop
for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
# continue is used to move to next iteration (step) - moves to next iteration of loop
for i in range(1,11):
    if i % 2 == 1:
        continue
    print(i)
```

```python
# pass does nothing 
for i in range(1,11):
    pass
    print(i)
```
