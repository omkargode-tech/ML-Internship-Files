# Part 1: Introduction to Decision Trees

---

## 1.1 What is a Decision Tree?

A **Decision Tree** is a supervised learning algorithm that makes predictions by asking a series of yes/no (or threshold) questions, forming a tree-like structure.

> 🧠 **Analogy — "20 Questions" game:**
> Think of how you'd guess an object in 20 questions. You start with broad questions ("Is it alive?"), narrow down based on answers, and eventually reach a specific guess. Decision Trees do exactly this — but they **learn** the best questions from data!

### Example: Should I play tennis?

```
                Outlook?
               /    |    \
            Sunny  Overcast  Rainy
             |       |        |
         Humidity?  PLAY   Windy?
          /    \              /    \
       High  Normal        Weak   Strong
        |      |            |       |
       NO    PLAY         PLAY     NO
```

Given the weather, you traverse the tree from the root, answer questions at each node, and reach a leaf with the decision.

### Formal Definition:

A decision tree is a **flowchart-like structure** where:
- **Internal node:** A test on a feature (e.g., "Age < 30?")
- **Branch:** The outcome of the test (e.g., True/False)
- **Leaf node:** A prediction (class label or value)
- **Root:** The topmost node (first question)

To predict, you start at the root and traverse down following the answers until you reach a leaf.

## 1.2 Why Decision Trees?

### 1. 🏗️ **Foundation of Modern ML**
Decision Trees are the building blocks of:
- **Random Forests** (ensemble of trees)
- **Gradient Boosting** (XGBoost, LightGBM, CatBoost)
- **AdaBoost**

### 2. 📖 **Highly Interpretable**
You can **draw** the tree and explain it to a non-technical stakeholder. This is huge for:
- Medicine (doctors need to trust the model)
- Finance (regulations require explanation)
- Business (executives need to understand decisions)

### 3. ⚡ **No Feature Scaling Needed**
Trees split on thresholds — they don't care if a feature is in [0,1] or [0, 1000000].

### 4. 🎯 **Handles Mixed Data Types**
Naturally handles:
- Continuous features (Age = 35)
- Categorical features (Color = "Red")
- Binary features (Is_Employed = True)

### 5. 🧠 **Mimics Human Decision-Making**
Trees are how humans often make decisions — a series of if-then rules.

### 6. 🔄 **Non-Parametric**
No assumption about data distribution. Trees can fit any shape (linear, non-linear, step-function, etc.).

## 1.3 Types of Decision Trees

| Type | Output | Example Algorithm |
|------|--------|-------------------|
| **Classification Tree** | Discrete class labels | ID3, C4.5, CART |
| **Regression Tree** | Continuous values | CART |

### Classification vs Regression Trees:

| | Classification | Regression |
|---|---|---|
| **Leaf value** | Majority class | Mean of samples |
| **Split criterion** | Gini, Entropy | MSE, MAE |
| **Prediction** | Class label | Number |

## 1.4 Famous Decision Tree Algorithms

| Algorithm | Year | Split Criterion | Features |
|-----------|------|-----------------|----------|

| **C4.5** | 1993 (Quinlan) | Gain Ratio | Categorical + Continuous |

| **CHAID** | 1980 (Kass) | Chi-squared | Multi-way splits |

> 📌 sklearn uses **CART** (Classification and Regression Trees), which produces **binary** trees.

## 1.5 Where Do Decision Trees Fit?

```
Machine Learning
└── Supervised Learning
    ├── Regression → Linear Regression, Regression Trees
    └── Classification → Logistic Regression, Classification Trees, Naive Bayes
```

Decision Trees can do **both** regression and classification — they're versatile!

# Part 2: Understanding Decision Trees

---

## 2.1 Anatomy of a Decision Tree

```
                    [ROOT] Age < 30?
                   /              \
                Yes                No
                 /                  \
        [INTERNAL] Income > 50k?    [LEAF] Don't Buy
              /        \
           Yes          No
            /            \
      [LEAF] Buy     [LEAF] Don't Buy
```

### Components:

| Component | Description |
|-----------|-------------|
| **Root node** | Top of the tree; first split |
| **Internal node** | A decision point (feature test) |
| **Branch/Edge** | Outcome of a test (True/False or value) |
| **Leaf node** | Terminal node with a prediction |
| **Depth** | Number of edges from root to a node |
| **Parent/Child** | Node above/below in the hierarchy |

## 2.2 How a Tree Makes Predictions

For a new input, traverse from the root:
1. At each internal node, test the specified feature.
2. Follow the branch corresponding to the test result.
3. Repeat until reaching a leaf.
4. The leaf gives the prediction (class label or value).

### Example:
Input: `Age=25, Income=60k`
- Start at root: `Age < 30?` → Yes → go left
- At `Income > 50k?` → Yes → go left
- Reach leaf: **Buy**

## 2.3 Tree Building (Training)

The training process **recursively** splits the data:

1. **Start** with all training data at the root.
2. **Find the best split** (feature + threshold) that maximally separates classes.
3. **Split** the data into subsets based on the chosen test.
4. **Recurse** on each subset.
5. **Stop** when:
   - All samples in a node belong to the same class (pure node)
   - No more features to split on
   - Maximum depth reached
   - Minimum samples per leaf reached

### The Key Question:
> How do we find the "best" split? This is where **impurity measures** come in (Part 4-5).

## 2.4 Classification vs Regression Trees

### Classification Tree:
- **Leaf value:** Most common class in the leaf (majority vote)
- **Split criterion:** Gini impurity or Entropy (maximize information gain)
- **Prediction:** Class label

### Regression Tree:
- **Leaf value:** Mean of target values in the leaf
- **Split criterion:** Minimize MSE (variance) in the leaves
- **Prediction:** Continuous number

### Example:
```
Regression Tree:                Classification Tree:
   Age < 30?                       Age < 30?
   /        \                      /        \
  Yes        No                   Yes        No
  |          |                    |          |
[Mean=25] [Mean=45]            [Class=A]  [Class=B]
```

---

### 📝 Part 2 Summary

| Concept | Key Idea |
|---------|----------|
| Tree structure | Root → internal → leaves |
| Prediction | Traverse from root following tests |
| Building | Recursively split to reduce impurity |
| Classification leaf | Majority class |
| Regression leaf | Mean value |

### ✅ Key Takeaways
- Trees are built top-down, recursively.
- Each split partitions data into purer subsets.
- Classification uses majority vote; regression uses mean.
- The "best split" maximizes purity gain.

### ⚠️ Common Mistakes
- Confusing tree depth with number of nodes.
- Forgetting that CART produces **binary** trees.
- Not understanding the recursive nature of tree building.

# Part 3: Intuition Behind Decision Trees

---

## 3.1 The Core Idea

> **A decision tree finds the questions that best separate the classes (or reduce target variance).**

### The "Best Question" Principle:
At each node, the tree asks: *"If I split on feature X with threshold t, how much does this reduce uncertainty?"*

The split that reduces uncertainty the most is chosen.

## 3.2 Example: Should We Approve a Loan?

Imagine a dataset of loan applicants:

| Age | Income | Credit_Score | Default? |
|-----|--------|--------------|----------|
| 25  | 30k    | Fair         | Yes      |
| 35  | 60k    | Good         | No       |
| 45  | 80k    | Excellent    | No       |
| 23  | 25k    | Poor         | Yes      |
| 55  | 100k   | Good         | No       |

### Tree's Logic:
1. **First question:** "Credit_Score = Poor?" → Best separator!
   - Poor → Default (mostly)
   - Not Poor → No Default (mostly)
2. **Next question (in "Not Poor" branch):** "Income < 50k?"
   - < 50k → Maybe Default
   - ≥ 50k → No Default

The tree found that **Credit_Score** is the most informative feature first.

## 3.3 Visual Intuition: Decision Boundaries

Unlike Logistic Regression (linear boundary), Decision Trees create **axis-aligned** (rectangular) boundaries.


````python
# ============================================================
# Visualize: Decision Tree vs Logistic Regression Boundaries
# ============================================================
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

np.random.seed(42)
X, y = make_classification(n_samples=200, n_features=2, n_redundant=0, 
                            n_informative=2, random_state=42, n_clusters_per_class=1)

# Train both models
tree = DecisionTreeClassifier(max_depth=3, random_state=42)
log_reg = LogisticRegression(random_state=42)
tree.fit(X, y)
log_reg.fit(X, y)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))

for ax, model, title in [(axes[0], log_reg, 'Logistic Regression (linear boundary)'), 
                          (axes[1], tree, 'Decision Tree (axis-aligned boundary)')]:
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
    ax.contour(xx, yy, Z, levels=[0.5], colors='black', linewidths=2)
    ax.scatter(X[y==0, 0], X[y==0, 1], color='purple', s=30, edgecolor='k', label='Class 0')
    ax.scatter(X[y==1, 0], X[y==1, 1], color='green', s=30, edgecolor='k', label='Class 1')
    acc = model.score(X, y)
    ax.set_title(f'{title}\n(Acc: {acc:.3f})', fontweight='bold')
    ax.legend()

plt.tight_layout()
plt.show()
print("💡 Decision Trees create rectangular (staircase) boundaries — each split is axis-aligned!")
````

## 3.4 The "20 Questions" Analogy

Imagine you're trying to identify an animal:

1. **Q1: Is it a mammal?** (Yes)
   → narrows from all animals to mammals
2. **Q2: Is it larger than a human?** (No)
   → narrows to small mammals
3. **Q3: Does it bark?** (Yes)
   → narrows to dogs

Each question **maximally reduces uncertainty** about the identity. This is exactly how a decision tree decides which feature to split on!

## 3.5 Purity and Impurity

### Pure Node:
All samples belong to the same class. Example: 100 samples, all class "Spam".

### Impure Node:
Samples are mixed. Example: 50 "Spam", 50 "Ham".

### Goal:
Make nodes as **pure** as possible through splits.

```
Before split (impure):     After split (purer):
[50 Spam, 50 Ham]    →    [45 Spam, 5 Ham]   (mostly Spam)
                        [5 Spam, 45 Ham]    (mostly Ham)
```

The split **reduced impurity** — we're closer to pure nodes!

## 3.6 Information Gain

> **Information Gain** = how much impurity decreases after a split.

$$\text{IG} = \text{Impurity}_{\text{parent}} - \sum_{\text{children}} \frac{n_{\text{child}}}{n_{\text{parent}}} \cdot \text{Impurity}_{\text{child}}$$

The tree picks the split that **maximizes** Information Gain.

## 3.7 Recursive Partitioning

The tree building is **recursive**:
1. Split the data into 2 subsets.
2. For each subset, find the best split.
3. Repeat until stopping criterion.

```
                 [All Data]
                /          \
         [Subset 1]     [Subset 2]
          /     \         /     \
       [L]    [R]      [L]    [R]
       ...    ...      ...    ...
```

This creates the tree structure automatically!

---

### 📝 Part 3 Summary

| Concept | Key Idea |
|---------|----------|
| Best split | Maximally separates classes (reduces impurity) |
| Decision boundary | Axis-aligned (rectangular) |
| Pure node | All same class |
| Information gain | Reduction in impurity from a split |
| Recursive | Repeatedly split subsets |

### ✅ Key Takeaways
- Trees ask the most informative questions first.
- Boundaries are axis-aligned (staircase pattern).
- Goal: reduce impurity at each split.
- Building is recursive partitioning.

### ⚠️ Common Mistakes
- Expecting smooth/linear boundaries from trees.
- Letting trees split until perfectly pure (overfitting).
- Not realizing each split affects only one subset.

# Part 4: Mathematics of Decision Trees

---

## 4.1 Impurity Measures

The math of decision trees centers on **measuring node impurity**. The three main measures:

### 1. Gini Impurity (CART default)

$$\text{Gini}(t) = 1 - \sum_{i=1}^{K} p_i^2$$

where:
- $K$ = number of classes
- $p_i$ = proportion of class $i$ in node $t$

**Interpretation:** Probability of misclassifying a randomly chosen sample if labeled according to the class distribution in the node.

**Range:** [0, 1 - 1/K] (0 = pure, higher = more impure)

### 2. Entropy (Information Theory)

$$\text{Entropy}(t) = -\sum_{i=1}^{K} p_i \log_2(p_i)$$

**Interpretation:** Average number of bits needed to encode the class of a randomly chosen sample.

**Range:** [0, log₂(K)] (0 = pure, log₂(K) = maximally impure)

### 3. Classification Error

$$\text{Error}(t) = 1 - \max_i p_i$$

**Interpretation:** Probability of error if we predict the majority class.

**Range:** [0, 1 - 1/K]

## 4.2 Comparing Impurity Measures

| Measure | Formula | Smooth? | Default in |
|---------|---------|---------|------------|
| Gini | $1 - \sum p_i^2$ | Yes | sklearn CART |
| Entropy | $-\sum p_i \log_2 p_i$ | Yes | ID3, C4.5 |
| Error | $1 - \max p_i$ | No (not differentiable) | Pruning |

> 💡 Gini and Entropy usually give **very similar trees**. Gini is faster (no log computation).

## 4.3 Information Gain

$$\text{IG}(\text{split}) = \text{Impurity}(\text{parent}) - \sum_{j} \frac{n_j}{n} \text{Impurity}(\text{child}_j)$$

The tree chooses the split that **maximizes** IG.

### Example:
Parent node: 50 Spam, 50 Ham → Gini = 0.5

Split A: [40 Spam, 10 Ham] and [10 Spam, 40 Ham]
- Gini(child1) = 1 - (0.8² + 0.2²) = 0.32
- Gini(child2) = 1 - (0.2² + 0.8²) = 0.32
- IG = 0.5 - (0.5 × 0.32 + 0.5 × 0.32) = 0.5 - 0.32 = **0.18**

Split B: [25 Spam, 25 Ham] and [25 Spam, 25 Ham]
- Gini(child1) = Gini(child2) = 0.5
- IG = 0.5 - 0.5 = **0.0** (no information gained)

→ Tree picks Split A (higher IG).

## 4.4 Gini Gain (for CART)

Same as Information Gain, but using Gini impurity:
$$\Delta\text{Gini} = \text{Gini}(\text{parent}) - \sum_j \frac{n_j}{n}\text{Gini}(\text{child}_j)$$

## 4.5 Regression: Variance Reduction

For regression trees, we minimize **Mean Squared Error** (variance):

$$\text{MSE}(t) = \frac{1}{n_t}\sum_{i \in t}(y_i - \bar{y}_t)^2$$

The split that **maximizes variance reduction**:
$$\Delta\text{MSE} = \text{MSE}(\text{parent}) - \sum_j \frac{n_j}{n}\text{MSE}(\text{child}_j)$$

## 4.6 The Splitting Problem

For continuous features, we need to find the best threshold $t$:
- Sort the feature values.
- Try midpoints between consecutive values as candidate thresholds.
- Pick the one with highest IG.

For $n$ samples, this is $O(n \log n)$ per feature (due to sorting).

### Computational Complexity:
- Building: $O(n \cdot p \cdot \log n \cdot d)$ where $d$ = depth
- Prediction: $O(d)$ per sample


````python
# ============================================================
# Visualize: Gini vs Entropy vs Classification Error
# ============================================================
# For binary classification: vary p (proportion of class 1)
p = np.linspace(0.001, 0.999, 200)

gini = 1 - (p**2 + (1-p)**2)
entropy = -p * np.log2(p) - (1-p) * np.log2(1-p)
error = 1 - np.maximum(p, 1-p)

plt.figure(figsize=(10, 6))
plt.plot(p, gini, 'steelblue', linewidth=2.5, label='Gini')
plt.plot(p, entropy, 'coral', linewidth=2.5, label='Entropy (scaled to [0,1])')
plt.plot(p, error, 'green', linewidth=2.5, label='Classification Error')
plt.xlabel('p (proportion of class 1)')
plt.ylabel('Impurity')
plt.title('Impurity Measures for Binary Classification', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("💡 All three measures are maximized at p=0.5 (most impure) and 0 at p=0 or 1 (pure).")
print("   Gini and Entropy are very similar — both are smooth and differentiable.")
````

# Part 5: Impurity Measures (Cost Function)

---

## 5.1 The Tree's "Cost Function"

Decision Trees don't have a single global cost function like MSE or cross-entropy. Instead, they greedily minimize **impurity at each node**.

### Greedy approach:
- At each node, find the split that maximizes information gain.
- Don't look ahead — just take the best split now.
- This is **greedy** and **not globally optimal**, but works well in practice.

### Why greedy?
- Finding the globally optimal tree is **NP-hard** (combinatorial explosion).
- Greedy is fast and gives good (not optimal) trees.

## 5.2 Detailed: Gini Impurity

$$\text{Gini}(t) = 1 - \sum_{i=1}^{K} p_i^2$$

### Properties:
- 0 when node is pure (one class).
- Maximum when classes are equally distributed.
- For binary: max Gini = 0.5 (at p=0.5).
- For K-class: max Gini = $1 - 1/K$.

### Why Gini?
- **Computationally cheap** (no logarithms).
- **Differentiable** (good for optimization).
- **Similar performance** to entropy.

## 5.3 Detailed: Entropy

$$\text{Entropy}(t) = -\sum_{i=1}^{K} p_i \log_2(p_i)$$

### Properties:
- 0 when node is pure.
- Maximum $\log_2(K)$ when classes are equally distributed.
- Comes from **information theory** (Shannon entropy).
- Slightly more expensive (log computation).

### Information Gain with Entropy:
$$\text{IG} = \text{Entropy}(\text{parent}) - \sum_j \frac{n_j}{n}\text{Entropy}(\text{child}_j)$$

This is the original ID3 algorithm.

## 5.4 Detailed: Regression Cost (MSE)

For regression trees:

$$\text{MSE}(t) = \frac{1}{n_t}\sum_{i \in t}(y_i - \bar{y}_t)^2$$

The split that maximizes:
$$\Delta\text{MSE} = \text{MSE}(\text{parent}) - \sum_j \frac{n_j}{n}\text{MSE}(\text{child}_j)$$

### Alternative: MAE
$$\text{MAE}(t) = \frac{1}{n_t}\sum_{i \in t}|y_i - \text{median}(t)|$$

Less common, more robust to outliers, but harder to optimize.

## 5.5 When to Use Which?

| Task | Impurity | Algorithm |
|------|----------|-----------|
| Classification (fast) | Gini | CART (sklearn default) |
| Classification (info-theory) | Entropy | ID3/C4.5 style |
| Regression | MSE | CART |
| Robust regression | MAE | Custom

> 💡 In practice, **Gini vs Entropy rarely matters** — they produce nearly identical trees. Pick one and move on.

# Part 6: Tree Building Algorithm

---

## 6.1 The CART Algorithm (sklearn)

```
BuildTree(Data, depth=0):
    IF stopping_criterion_met(Data, depth):
        return Leaf(predicted_class_or_value)
    
    best_split = FindBestSplit(Data)
    IF best_split is None:
        return Leaf(...)
    
    left_data, right_data = Split(Data, best_split)
    left_child = BuildTree(left_data, depth+1)
    right_child = BuildTree(right_data, depth+1)
    
    return Node(best_split, left_child, right_child)
```

## 6.2 FindBestSplit

```
FindBestSplit(Data):
    best_gain = -inf
    best_split = None
    
    FOR each feature f:
        FOR each threshold t (candidate):
            left, right = Split(Data, f, t)
            gain = Impurity(Data) - weighted_avg(Impurity(left), Impurity(right))
            IF gain > best_gain:
                best_gain = gain
                best_split = (f, t)
    
    return best_split
```

### Candidate thresholds:
- For continuous features: midpoints between consecutive sorted values.
- For categorical: try each category (or use one-hot encoding).

## 6.3 Time Complexity

- Per split: $O(p \cdot n \log n)$ (sort each feature, evaluate splits)
- Total: $O(p \cdot n \log n \cdot d)$ where $d$ = depth

## 6.4 Stopping Criteria

The recursion stops when:
1. **Pure node:** All samples same class.
2. **No features left:** All features exhausted.
3. **Max depth:** `max_depth` reached.
4. **Min samples per split:** `min_samples_split` not met.
5. **Min samples per leaf:** `min_samples_leaf` not met.
6. **No improvement:** Information gain below threshold.

These are **pre-pruning** parameters (Part 14).

---

### 📝 Part 6 Summary

| Concept | Key Idea |
|---------|----------|
| Algorithm | Recursive greedy splitting |
| FindBestSplit | Try all features × thresholds |
| Complexity | $O(p \cdot n \log n \cdot d)$ |
| Stopping | Pure node, max depth, min samples |

### ✅ Key Takeaways
- CART builds trees recursively, top-down.
- Each split is chosen greedily (best now, not globally).
- Stopping criteria prevent infinite recursion.
- Complexity scales with features, samples, and depth.


# Part 7: Splitting Criteria & Stopping Rules

---

## 7.1 Splitting Rules

### For Continuous Features:
- Try all midpoints between sorted values.
- Example: feature values [1, 3, 7, 10] → thresholds [2, 5, 8.5].

### For Categorical Features:
- **Binary split (CART):** "Feature == category_A?" vs "Feature != category_A"
- **Multiway split (ID3/C4.5):** One branch per category (sklearn doesn't support this — use one-hot encoding).

## 7.2 Pre-Pruning (Stopping Rules)

sklearn's `DecisionTreeClassifier` parameters:

| Parameter | Default | Description |
|-----------|---------|-------------|
| `max_depth` | None | Maximum depth of tree |
| `min_samples_split` | 2 | Min samples to split a node |
| `min_samples_leaf` | 1 | Min samples in a leaf |
| `max_features` | None | Max features to consider per split |
| `max_leaf_nodes` | None | Max number of leaves |
| `min_impurity_decrease` | 0 | Min impurity decrease required to split |

### Effect of each:

| Parameter | Increase → | Effect |
|-----------|-----------|--------|
| `max_depth` | ↑ | Deeper tree, more overfitting |
| `min_samples_split` | ↑ | Fewer splits, less overfitting |
| `min_samples_leaf` | ↑ | Smoother, less overfitting |
| `max_features` | ↓ | More random, less overfitting (used in Random Forest) |

## 7.3 Post-Pruning (Cost-Complexity)

After building the full tree, prune back branches that don't improve validation performance.

### Cost-Complexity Pruning (CART):
$$R_\alpha(T) = R(T) + \alpha |T|$$

where:
- $R(T)$ = training error of tree $T$
- $|T|$ = number of leaves
- $\alpha$ = complexity penalty

Larger $\alpha$ → more pruning. sklearn exposes this via `ccp_alpha`.

---

### 📝 Part 7 Summary

| Concept | Key Idea |
|---------|----------|
| Continuous splits | Try midpoints |
| Categorical splits | Binary (CART) or one-hot |
| Pre-pruning | Stop early (max_depth, etc.) |
| Post-pruning | Prune after building (ccp_alpha) |

### ✅ Key Takeaways
- Pre-pruning stops tree growth early.
- Post-pruning removes branches after building.
- `max_depth` and `min_samples_leaf` are the most important parameters.
- Use cross-validation to tune pruning parameters.


# Part 8: From-Scratch Implementation

---

## 8.1 Build CART from Scratch

We'll implement a simple CART classifier using Gini impurity.


````python
# ============================================================
# Decision Tree from Scratch (CART Classifier)
# ============================================================

class DecisionTreeScratch:
    """
    Binary CART Decision Tree Classifier implemented from scratch.
    Uses Gini impurity for splitting.
    """
    def __init__(self, max_depth=5, min_samples_split=2, min_samples_leaf=1):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.min_samples_leaf = min_samples_leaf
        self.tree = None
        self.n_classes = None
        self.feature_importances_ = None
    
    def _gini(self, y):
        """Compute Gini impurity of labels y."""
        if len(y) == 0:
            return 0
        _, counts = np.unique(y, return_counts=True)
        p = counts / len(y)
        return 1 - np.sum(p ** 2)
    
    def _best_split(self, X, y):
        """Find the best feature and threshold to split on."""
        n_samples, n_features = X.shape
        if n_samples < self.min_samples_split:
            return None, None, 0
        
        parent_gini = self._gini(y)
        best_gain = 0
        best_feature = None
        best_threshold = None
        
        for feature in range(n_features):
            # Get sorted unique values as candidate thresholds
            thresholds = np.unique(X[:, feature])
            if len(thresholds) < 2:
                continue
            
            # Use midpoints between consecutive sorted values
            thresholds = (thresholds[:-1] + thresholds[1:]) / 2
            
            for threshold in thresholds:
                left_mask = X[:, feature] <= threshold
                right_mask = ~left_mask
                
                n_left = np.sum(left_mask)
                n_right = np.sum(right_mask)
                
                # Check min_samples_leaf
                if n_left < self.min_samples_leaf or n_right < self.min_samples_leaf:
                    continue
                
                gini_left = self._gini(y[left_mask])
                gini_right = self._gini(y[right_mask])
                
                # Weighted child impurity
                weighted_gini = (n_left / n_samples) * gini_left + (n_right / n_samples) * gini_right
                
                # Information gain
                gain = parent_gini - weighted_gini
                
                if gain > best_gain:
                    best_gain = gain
                    best_feature = feature
                    best_threshold = threshold
        
        return best_feature, best_threshold, best_gain
    
    def _build_tree(self, X, y, depth=0):
        """Recursively build the tree."""
        n_samples = len(y)
        
        # Count feature usage for importance
        leaf_value = np.bincount(y, minlength=self.n_classes)
        
        # Stopping criteria
        if (depth >= self.max_depth or 
            n_samples < self.min_samples_split or
            self._gini(y) == 0):
            return {'type': 'leaf', 'prediction': np.argmax(leaf_value), 'samples': n_samples, 'class_counts': leaf_value}
        
        # Find best split
        feature, threshold, gain = self._best_split(X, y)
        
        if feature is None or gain == 0:
            return {'type': 'leaf', 'prediction': np.argmax(leaf_value), 'samples': n_samples, 'class_counts': leaf_value}
        
        # Track feature importance
        self._feature_usage[feature] = self._feature_usage.get(feature, 0) + gain * n_samples
        
        # Split
        left_mask = X[:, feature] <= threshold
        left = self._build_tree(X[left_mask], y[left_mask], depth + 1)
        right = self._build_tree(X[~left_mask], y[~left_mask], depth + 1)
        
        return {
            'type': 'node',
            'feature': feature,
            'threshold': threshold,
            'gain': gain,
            'left': left,
            'right': right,
            'samples': n_samples
        }
    
    def _predict_one(self, x, node):
        """Predict for a single sample by traversing the tree."""
        while node['type'] == 'node':
            if x[node['feature']] <= node['threshold']:
                node = node['left']
            else:
                node = node['right']
        return node['prediction']
    
    def fit(self, X, y):
        """Train the decision tree."""
        X = np.array(X, dtype=float)
        y = np.array(y, dtype=int)
        self.n_classes = len(np.unique(y))
        self._feature_usage = {}
        self.tree = self._build_tree(X, y, depth=0)
        
        # Compute feature importances (normalized)
        total = sum(self._feature_usage.values())
        if total > 0:
            self.feature_importances_ = np.array([
                self._feature_usage.get(i, 0) / total for i in range(X.shape[1])
            ])
        else:
            self.feature_importances_ = np.zeros(X.shape[1])
        
        return self
    
    def predict(self, X):
        """Predict class labels."""
        X = np.array(X, dtype=float)
        return np.array([self._predict_one(x, self.tree) for x in X])
    
    def score(self, X, y):
        """Compute accuracy."""
        return np.mean(self.predict(X) == np.array(y))
    
    def print_tree(self, node=None, depth=0, feature_names=None):
        """Print the tree structure."""
        if node is None:
            node = self.tree
        indent = "  " * depth
        if node['type'] == 'leaf':
            print(f"{indent}→ Predict class {node['prediction']} ({node['samples']} samples)")
        else:
            fname = feature_names[node['feature']] if feature_names else f"X[{node['feature']}]"
            print(f"{indent}{fname} <= {node['threshold']:.3f}? (gain={node['gain']:.4f}, {node['samples']} samples)")
            self.print_tree(node['left'], depth + 1, feature_names)
            self.print_tree(node['right'], depth + 1, feature_names)

print("✅ DecisionTreeScratch class defined!")
````

````python
# ============================================================
# Test the from-scratch Decision Tree
# ============================================================
np.random.seed(42)

# Use Iris dataset (first 2 features for simplicity)
X_test_tree = iris.data[:100, :2]  # First 100 samples, 2 features
y_test_tree = iris.target[:100]

# Train
scratch_tree = DecisionTreeScratch(max_depth=3, min_samples_split=5)
scratch_tree.fit(X_test_tree, y_test_tree)

# Print tree
print("🌳 Trained Decision Tree:")
scratch_tree.print_tree(feature_names=iris.feature_names[:2])

# Evaluate
accuracy = scratch_tree.score(X_test_tree, y_test_tree)
print(f"\n📊 Training accuracy: {accuracy:.4f}")
print(f"   Feature importances: {scratch_tree.feature_importances_}")

# Visualize decision boundary
plt.figure(figsize=(8, 6))
x_min, x_max = X_test_tree[:, 0].min() - 0.5, X_test_tree[:, 0].max() + 0.5
y_min, y_max = X_test_tree[:, 1].min() - 0.5, X_test_tree[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 100), np.linspace(y_min, y_max, 100))
Z = scratch_tree.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
plt.scatter(X_test_tree[y_test_tree==0, 0], X_test_tree[y_test_tree==0, 1], color='purple', s=40, edgecolor='k', label=iris.target_names[0])
plt.scatter(X_test_tree[y_test_tree==1, 0], X_test_tree[y_test_tree==1, 1], color='green', s=40, edgecolor='k', label=iris.target_names[1])
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title('🌳 From-Scratch Decision Tree Boundary', fontweight='bold')
plt.legend()
plt.tight_layout()
plt.show()
````

## 8.2 Understanding the Code

### `_gini(y)`

### `_best_split(X, y)`

### `_predict_one(x, node)`
Traverses the tree for a single sample until reaching a leaf.

### `fit`, `predict`, `score`
Standard sklearn-like API.

# Part 11: Data Preprocessing for Trees

---

## 11.1 What Trees DON'T Need

Decision Trees are low-maintenance:
- ❌ **No feature scaling** (thresholds are scale-invariant)
- ❌ **No normalization**
- ❌ **No monotonic transformations** (order is preserved)
- ❌ **No dummy variable trap** (one-hot is fine but not required for binary trees)

## 11.2 What Trees DO Need

- ✅ **Handle missing values** (sklearn doesn't natively, impute first)
- ✅ **Encode categorical features** (one-hot or ordinal)
- ✅ **Handle class imbalance** (`class_weight='balanced'`)
- ✅ **Train/test split** (still needed!)

## 11.3 Categorical Encoding for Trees

### Option 1: One-Hot Encoding
```python
pd.get_dummies(df, columns=['color'])
```
- Pro: Works with all algorithms
- Con: Increases dimensionality

### Option 2: Ordinal Encoding (for trees only!)
```python
from sklearn.preprocessing import OrdinalEncoder
```
- Pro: No dimensionality increase
- Con: Trees may treat categories ordinally (which they're not)
- Works okay for trees because they find thresholds

> 💡 sklearn trees work best with one-hot encoding for nominal categories.

## 11.4 Handling Missing Values

sklearn trees don't natively handle missing values (unlike XGBoost/LightGBM). Options:
1. **Impute** (mean/median/mode)
2. **Add missing indicator** (`MissingIndicator`)
3. **Use a different library** (XGBoost, LightGBM handle NaN natively)


````python
# ============================================================
# Demo: Trees Don't Need Scaling
# ============================================================
from sklearn.preprocessing import StandardScaler

# Use Breast Cancer (features have very different scales)
X_tr_d, X_te_d, y_tr_d, y_te_d = train_test_split(cancer.data, cancer.target, test_size=0.2, random_state=42, stratify=cancer.target)

# Without scaling
tree_no_scale = DecisionTreeClassifier(max_depth=5, random_state=42)
tree_no_scale.fit(X_tr_d, y_tr_d)
acc_no_scale = tree_no_scale.score(X_te_d, y_te_d)

# With scaling (should give identical results!)
scaler = StandardScaler()
X_tr_s = scaler.fit_transform(X_tr_d)
X_te_s = scaler.transform(X_te_d)
tree_scaled = DecisionTreeClassifier(max_depth=5, random_state=42)
tree_scaled.fit(X_tr_s, y_tr_d)
acc_scaled = tree_scaled.score(X_te_s, y_te_d)

print(f"📊 Decision Trees — Scaling Effect:")
print(f"   Without scaling: accuracy = {acc_no_scale:.4f}")
print(f"   With scaling:    accuracy = {acc_scaled:.4f}")
print(f"   → IDENTICAL! Trees are scale-invariant. 🎉")
````

# Part 12: Properties & Assumptions of Decision Trees

---

## 12.1 Key Properties

| Property | Description |
|----------|-------------|
| **Non-parametric** | No distributional assumptions |
| **Non-linear** | Can fit any shape |
| **Axis-aligned splits** | Boundaries are rectangular |
| **Greedy** | Locally optimal, not globally |
| **High variance** | Small data changes → very different trees |
| **Interpretable** | Can be visualized |

## 12.2 The "No Assumptions" Advantage

Unlike Linear/Logistic Regression, trees make **no assumptions** about:
- Linearity
- Normality
- Homoscedasticity
- Independence

This makes them very flexible — but also prone to overfitting.

## 12.3 The High Variance Problem

> **Small changes in training data → very different trees.**

This is why single trees are rarely used in production — ensembles (Random Forest, Gradient Boosting) reduce variance by averaging many trees.

## 12.4 Bias-Variance in Trees

| Tree depth | Bias | Variance |
|-----------|------|----------|
| Shallow (depth=1) | High (underfit) | Low |
| Medium | Medium | Medium |
| Deep (no limit) | Low (overfit) | High |

The goal is to find the sweet spot via pruning.

## 12.5 Extrapolation Issue

Decision Trees **cannot extrapolate**:
- For regression, predictions are bounded by training data range.
- A tree trained on house prices $100k-$500k will never predict $600k+.

> ⚠️ This is a major limitation for time-series forecasting!

---

### 📝 Part 12 Summary

| Property | Implication |
|----------|-------------|
| Non-parametric | Flexible, no assumptions |
| High variance | Use ensembles |
| Axis-aligned | Can't fit diagonal boundaries well |
| No extrapolation | Bad for forecasting |
| Greedy | Not globally optimal |

### ✅ Key Takeaways
- Trees make no distributional assumptions.
- Single trees have high variance — use ensembles.
- Trees can't extrapolate beyond training data range.
- Axis-aligned boundaries are a limitation (diagonal boundaries need many splits).

# Part 14: Pruning & Regularization

---

## 14.1 Why Prune?

An unpruned tree will:
- Grow until every leaf is pure.
- Memorize training data (100% train accuracy).
- Generalize poorly (high test error).

This is the **classic overfitting problem** for trees.

## 14.2 Pre-Pruning (Early Stopping)

Stop the tree from growing too deep:

| Parameter | Effect |
|-----------|--------|
| `max_depth=5` | Limit tree depth |
| `min_samples_split=10` | Don't split nodes with <10 samples |
| `min_samples_leaf=5` | Each leaf must have ≥5 samples |
| `max_leaf_nodes=20` | Limit number of leaves |
| `min_impurity_decrease=0.01` | Only split if meaningful |

## 14.3 Post-Pruning (Cost-Complexity)

Build full tree, then prune back:

$$R_\alpha(T) = R(T) + \alpha \cdot |T|$$

- $R(T)$ = training error
- $|T|$ = number of leaves
- $\alpha$ (`ccp_alpha` in sklearn) = complexity penalty

Larger $\alpha$ → more pruning.

## 14.4 Finding the Best Pruning Level

Use cross-validation:


````python
# ============================================================
# Pruning Demo: Effect of max_depth
# ============================================================
depths = range(1, 20)
train_scores = []
test_scores = []

for d in depths:
    t = DecisionTreeClassifier(max_depth=d, random_state=42)
    t.fit(X_tr_d, y_tr_d)
    train_scores.append(t.score(X_tr_d, y_tr_d))
    test_scores.append(t.score(X_te_d, y_te_d))

plt.figure(figsize=(10, 6))
plt.plot(depths, train_scores, 'b-o', linewidth=2, markersize=6, label='Train accuracy')
plt.plot(depths, test_scores, 'r-o', linewidth=2, markersize=6, label='Test accuracy')
plt.xlabel('Max Depth')
plt.ylabel('Accuracy')
plt.title('🌳 Effect of Tree Depth on Overfitting', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

best_depth = depths[np.argmax(test_scores)]
print(f"✅ Best max_depth: {best_depth} (test accuracy: {max(test_scores):.4f})")
print(f"   At depth=1: train={train_scores[0]:.4f}, test={test_scores[0]:.4f} (underfit)")
print(f"   At depth=15: train={train_scores[14]:.4f}, test={test_scores[14]:.4f} (overfit)")
````

````python
# ============================================================
# Cost-Complexity Pruning with ccp_alpha
# ============================================================
# Get pruning path
tree_full = DecisionTreeClassifier(random_state=42)
tree_full.fit(X_tr_d, y_tr_d)
path = tree_full.cost_complexity_pruning_path(X_tr_d, y_tr_d)
ccp_alphas = path.ccp_alphas

# Train trees for different ccp_alpha values
train_acc = []
test_acc = []
for alpha in ccp_alphas:
    t = DecisionTreeClassifier(ccp_alpha=alpha, random_state=42)
    t.fit(X_tr_d, y_tr_d)
    train_acc.append(t.score(X_tr_d, y_tr_d))
    test_acc.append(t.score(X_te_d, y_te_d))

plt.figure(figsize=(10, 6))
plt.plot(ccp_alphas, train_acc, 'b-o', linewidth=2, markersize=4, label='Train accuracy')
plt.plot(ccp_alphas, test_acc, 'r-o', linewidth=2, markersize=4, label='Test accuracy')
plt.xlabel('ccp_alpha (complexity penalty)')
plt.ylabel('Accuracy')
plt.title('🌳 Cost-Complexity Pruning', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xscale('log')
plt.tight_layout()
plt.show()

best_alpha = ccp_alphas[np.argmax(test_acc)]
print(f"✅ Best ccp_alpha: {best_alpha:.6f} (test accuracy: {max(test_acc):.4f})")
print("💡 As ccp_alpha increases, tree is pruned more — prevents overfitting!")
````

---

### 📝 Part 14 Summary

| Method | Description |
|--------|-------------|
| Pre-pruning | Stop growth early (max_depth, min_samples_leaf) |
| Post-pruning | Prune after building (ccp_alpha) |
| Cross-validation | Find optimal pruning level |

### ✅ Key Takeaways
- Unpruned trees overfit severely.
- Pre-pruning is simpler; post-pruning is more principled.
- Always use cross-validation to tune pruning parameters.
- The "sweet spot" balances bias and variance.


# Part 15: Feature Importance & Interpretability

---

## 15.1 What is Feature Importance?

A score for each feature indicating how useful it was for splitting.

### sklearn's computation:
$$\text{Importance}(f) = \sum_{\text{nodes splitting on } f} \frac{n_{\text{node}}}{n_{\text{total}}} \cdot \Delta\text{Impurity}$$

Then normalized to sum to 1.

### Interpretation:
- Higher = more important
- Sum to 1 across all features
- Many features often have 0 importance (not used in any split)

## 15.2 Advantages
- Easy to compute (built-in)
- Intuitive
- Feature selection (drop 0-importance features)

## 15.3 Limitations
- **Biased toward high-cardinality features** (features with many unique values get more chances to split)
- **Doesn't capture feature interactions** explicitly
- **Permutation importance** is more reliable

## 15.4 Permutation Importance (Better!)

Shuffle a feature's values and measure performance drop:
```
For each feature f:
    1. Note baseline accuracy
    2. Shuffle f's values (break relationship with target)
    3. Measure new accuracy
    4. Importance = baseline - shuffled
```


````python
# ============================================================
# Feature Importance Comparison
# ============================================================
from sklearn.inspection import permutation_importance

# Train tree
tree_fi = DecisionTreeClassifier(max_depth=5, random_state=42)
tree_fi.fit(X_tr_d, y_tr_d)

# Built-in feature importance
builtin_imp = tree_fi.feature_importances_

# Permutation importance (more reliable)
perm_result = permutation_importance(tree_fi, X_te_d, y_te_d, n_repeats=10, random_state=42)
perm_imp = perm_result.importances_mean

# Compare
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Built-in
top10_builtin = np.argsort(builtin_imp)[-10:]
axes[0].barh(range(10), builtin_imp[top10_builtin], color='steelblue')
axes[0].set_yticks(range(10))
axes[0].set_yticklabels([cancer.feature_names[i][:20] for i in top10_builtin], fontsize=9)
axes[0].set_xlabel('Importance')
axes[0].set_title('Built-in Feature Importance (Gini)', fontweight='bold')

# Permutation
top10_perm = np.argsort(perm_imp)[-10:]
axes[1].barh(range(10), perm_imp[top10_perm], color='coral')
axes[1].set_yticks(range(10))
axes[1].set_yticklabels([cancer.feature_names[i][:20] for i in top10_perm], fontsize=9)
axes[1].set_xlabel('Importance (accuracy drop)')
axes[1].set_title('Permutation Importance', fontweight='bold')

plt.tight_layout()
plt.show()

print("💡 Permutation importance is often more reliable than built-in!")
print("   Built-in is biased toward high-cardinality features.")
````

---

### 📝 Part 15 Summary

| Method | Description | Bias |
|--------|-------------|------|
| Built-in (Gini) | Sum of impurity decreases | Biased toward high-cardinality |
| Permutation | Performance drop when shuffled | Less biased, more expensive |

### ✅ Key Takeaways
- Feature importance reveals which features drive decisions.
- Built-in importance is fast but biased.
- Permutation importance is more reliable but slower.
- Use feature importance for feature selection and interpretation.

## 16.1 Axis-Aligned Boundaries

Decision Trees create **rectangular** (staircase) decision boundaries:
- Each split is parallel to an axis (one feature at a time)
- Diagonal boundaries require many small splits

## 16.2 Regression Trees: Step Functions

Regression trees predict a **constant** (the mean) in each leaf → predictions are **step functions**.

## 16.3 Comparison with Other Models


````python
# ============================================================
# Compare Decision Boundaries: Tree vs Logistic vs Tree Regression
# ============================================================
np.random.seed(42)

# Classification: compare boundaries
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Use Iris first 2 features
X_viz = iris.data[:, :2]
y_viz = iris.target

models = [
    (DecisionTreeClassifier(max_depth=3, random_state=42), 'Tree depth=3'),
    (DecisionTreeClassifier(max_depth=10, random_state=42), 'Tree depth=10 (overfit)'),
    (DecisionTreeClassifier(max_depth=3, min_samples_leaf=10, random_state=42), 'Tree depth=3, min_leaf=10')
]

x_min, x_max = X_viz[:, 0].min() - 0.5, X_viz[:, 0].max() + 0.5
y_min, y_max = X_viz[:, 1].min() - 0.5, X_viz[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))

for ax, (model, title) in zip(axes, models):
    model.fit(X_viz, y_viz)
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.3, cmap='viridis')
    ax.scatter(X_viz[:, 0], X_viz[:, 1], c=y_viz, cmap='viridis', s=30, edgecolor='k')
    ax.set_title(f'{title}\n(Acc: {model.score(X_viz, y_viz):.3f})', fontweight='bold')
    ax.set_xlabel(iris.feature_names[0])
    ax.set_ylabel(iris.feature_names[1])

plt.tight_layout()
plt.show()
print("💡 Notice the staircase boundaries — deeper trees overfit (smaller rectangles)!")
````

# Part 19: Interview Questions (55+)

---

## 🟢 Easy Questions (1-15)

### Q1: What is a Decision Tree?
**Answer:** A supervised learning algorithm that makes predictions via a tree-like structure of if-then rules. Internal nodes test features, branches represent outcomes, leaves give predictions. Used for both classification and regression.

### Q2: What's the difference between classification and regression trees?
**Answer:**
- **Classification:** Leaves predict class labels (majority vote); split by Gini/Entropy.
- **Regression:** Leaves predict continuous values (mean); split by MSE/variance reduction.

### Q3: What is Gini impurity?
**Answer:** $\text{Gini} = 1 - \sum p_i^2$. Measures how often a randomly chosen sample would be misclassified if labeled according to the node's class distribution. 0 = pure, max = 1 - 1/K.

### Q4: What is entropy?
**Answer:** $\text{Entropy} = -\sum p_i \log_2 p_i$. From information theory; measures average bits needed to encode class. 0 = pure, max = log₂(K).

### Q5: What is information gain?
**Answer:** $\text{IG} = \text{Impurity}(\text{parent}) - \sum \frac{n_j}{n}\text{Impurity}(\text{child}_j)$. The reduction in impurity from a split. Trees maximize IG.

### Q6: Why don't decision trees need feature scaling?
**Answer:** Trees split on thresholds — they don't compare features directly or compute distances. The scale of a feature doesn't affect the split decision.

### Q7: What is pruning?
**Answer:** Removing branches to prevent overfitting. Two types:
- **Pre-pruning:** Stop growth early (max_depth, min_samples_leaf).
- **Post-pruning:** Build full tree, then remove branches (ccp_alpha).

### Q8: What is the depth of a tree?
**Answer:** The number of edges from root to deepest leaf. Deeper trees are more complex and prone to overfitting.

### Q9: What does `max_depth` control?
**Answer:** The maximum depth of the tree. Smaller = simpler tree (less overfitting). Larger = more complex (more overfitting risk).

### Q10: What's a leaf node?
**Answer:** A terminal node with no children. In classification, predicts the majority class. In regression, predicts the mean of target values.

### Q11: What is the CART algorithm?
**Answer:** Classification and Regression Trees (Breiman, 1984). Produces **binary** trees (each node has exactly 2 children). sklearn uses CART.

### Q12: What is one-hot encoding, and why use it for trees?
**Answer:** Converting categorical features to binary columns. Used because sklearn trees don't natively handle categorical features (only numeric thresholds).

### Q13: What does `feature_importances_` tell you?
**Answer:** How much each feature contributed to reducing impurity across all splits. Higher = more important. Sum to 1.

### Q14: Why are decision trees called non-parametric?
**Answer:** They don't assume a fixed functional form (unlike Linear Regression's $y = mx + c$). The complexity grows with data.

### Q15: What's a pure node?
**Answer:** A node where all samples belong to the same class. Gini = 0, Entropy = 0. Trees stop splitting on pure nodes.

---

## 🟡 Medium Questions (16-35)

### Q16: Gini vs Entropy — which to use?
**Answer:** They give nearly identical trees. Gini is faster (no log computation) and is sklearn's default. Entropy has information-theory interpretation. Pick either — doesn't matter much.

### Q17: Why are decision trees high-variance?
**Answer:** Small changes in training data can lead to completely different splits, producing very different trees. This is why ensembles (Random Forest) help — they average many trees.

### Q18: What is the bias-variance tradeoff in trees?
**Answer:**
- Shallow tree: high bias (underfit), low variance.
- Deep tree: low bias (fit anything), high variance (overfit).
- Tune `max_depth` to find the sweet spot.

### Q19: How do trees handle categorical features?
**Answer:** sklearn trees don't natively handle them — you must one-hot encode first. Some libraries (LightGBM, CatBoost) handle categoricals natively.

### Q20: How do trees handle missing values?
**Answer:** sklearn: doesn't handle natively (impute first). XGBoost/LightGBM: handle natively by sending missing samples to the "best" child learned during training.

### Q21: What is cost-complexity pruning?
**Answer:** Post-pruning method: $R_\alpha(T) = R(T) + \alpha|T|$. Penalizes tree size (number of leaves). Larger α → more pruning. sklearn exposes via `ccp_alpha`.

### Q22: Can decision trees extrapolate?
**Answer:** No! Trees predict constant values in leaves. For regression, predictions are bounded by training data range. This makes them bad for time-series forecasting beyond observed range.

### Q23: What is `min_samples_leaf`?
**Answer:** Minimum number of samples required in a leaf node. Higher = smoother tree (less overfitting). Setting to 5+ is common.

### Q24: How does a tree decide the best split for continuous features?
**Answer:** Sort feature values, try midpoints between consecutive values as thresholds, pick the one with highest information gain. O(n log n) per feature.

### Q25: Why are tree boundaries axis-aligned?
**Answer:** Each split tests ONE feature at a time (e.g., `x1 < 5`). This creates boundaries parallel to axes. Diagonal boundaries need many small axis-aligned splits.

### Q26: What is the "20 questions" analogy?
**Answer:** Decision Trees work like the 20 Questions game — ask the most informative question first, narrow down based on answers, reach a conclusion. Each "question" is a split.

### Q27: How do you handle class imbalance in trees?
**Answer:** Use `class_weight='balanced'` to penalize minority class errors more. Also: oversampling, undersampling, or adjusting the decision threshold.

### Q28: What is `max_features` and why use it?
**Answer:** Limits features considered per split. Used in Random Forests to add randomness (decorrelate trees). Default for trees: all features. For forests: sqrt(p) or log2(p).

### Q29: How does a regression tree predict?
**Answer:** Traverses tree to a leaf, predicts the **mean** of training target values in that leaf. Predictions are step functions (constant per leaf).

### Q30: What's the time complexity of building a tree?
**Answer:** $O(p \cdot n \log n \cdot d)$ where $p$ = features, $n$ = samples, $d$ = depth. Sorting dominates per-split cost.

### Q31: What is the difference between ID3, C4.5, and CART?
**Answer:**
- **ID3:** Entropy, categorical only, multiway splits.
- **C4.5:** Gain Ratio, categorical + continuous, handles missing values.
- **CART:** Gini/MSE, binary splits, both classification and regression (sklearn uses this).

### Q32: Why are single trees rarely used in production?
**Answer:** High variance — small data changes produce very different trees. Ensembles (Random Forest, Gradient Boosting) average many trees for stability and better performance.

### Q33: What is `predict_proba` for trees?
**Answer:** Returns the **class proportion** in the leaf each sample falls into. NOT well-calibrated (e.g., 0.8 doesn't mean 80% true probability). Use `CalibratedClassifierCV` if calibration matters.

### Q34: How do you tune a decision tree?
**Answer:** Use GridSearchCV or RandomizedSearchCV with cross-validation on:
- `max_depth`: [3, 5, 7, 10, None]
- `min_samples_split`: [2, 5, 10]
- `min_samples_leaf`: [1, 2, 5]
- `criterion`: ['gini', 'entropy']

### Q35: What are the advantages of decision trees?
**Answer:**
1. Interpretable (can visualize)
2. No scaling needed
3. Handles mixed data types
4. Non-parametric (no assumptions)
5. Feature importance built-in
6. Fast prediction (O(depth))

## 🔴 Hard Questions (36-55)

### Q36: Derive information gain for entropy.
**Answer:**
$$\text{IG} = H(\text{parent}) - \sum_j \frac{n_j}{n}H(\text{child}_j)$$
where $H(t) = -\sum p_i \log_2 p_i$. The tree picks the split maximizing IG.

### Q37: Why is finding the globally optimal tree NP-hard?
**Answer:** The number of possible trees grows exponentially with features and samples. Greedy (CART) is $O(p \cdot n \log n \cdot d)$, but optimal requires exploring all possible trees — combinatorial explosion.

### Q38: Explain why Gini and Entropy give similar trees.
**Answer:** Both are concave, maximized at uniform distribution, minimized at pure nodes. Their shapes are similar (Entropy slightly more peaked). The ranking of splits is nearly identical.

### Q39: What is the relationship between trees and rule-based systems?
**Answer:** A tree is equivalent to a set of if-then rules: each root-to-leaf path is one rule. Trees **learn** these rules from data, unlike expert systems where rules are hand-coded.

### Q40: How does cost-complexity pruning work mathematically?
**Answer:** Minimize $R_\alpha(T) = R(T) + \alpha|T|$ where $R(T)$ = training error, $|T|$ = leaves. For each α, there's an optimal subtree. As α increases, the optimal tree shrinks. Weakest-link pruning removes the subtree with smallest increase in $R$ per removed leaf.

### Q41: Why are decision tree probabilities not calibrated?
**Answer:** `predict_proba` returns class proportions in the leaf. A leaf with 8/10 positive samples gives 0.8, but this doesn't mean 80% true probability — small leaves have high variance. Use isotonic regression or Platt scaling to calibrate.

### Q42: How does feature importance bias toward high-cardinality features?
**Answer:** Features with many unique values have more candidate thresholds → more chances to split → more impurity reduction accumulated. Even random features can get high importance if they have many unique values. Permutation importance fixes this.

### Q43: Explain the connection between trees and entropy in information theory.
**Answer:** Entropy $H = -\sum p \log p$ measures uncertainty in bits. A split reduces entropy (information gain = bits of information learned). Trees greedily maximize information gain — they're literally extracting information from features.

### Q44: How do you handle categorical features with many categories?
**Answer:**
1. One-hot encoding (but creates many features)
2. Target encoding (replace category with mean target — risk of overfitting)
3. Embedding (learn a dense representation)
4. Use CatBoost/LightGBM which handle natively

### Q45: What is the difference between pre-pruning and post-pruning?
**Answer:**
- **Pre-pruning:** Stop growth early (max_depth, min_samples_leaf). Simple but may stop too early.
- **Post-pruning:** Build full tree, then remove branches (ccp_alpha). More principled — sees the full tree before deciding.

### Q46: Why can't decision trees extrapolate?
**Answer:** Each leaf predicts a constant (mean of training values in that leaf). For a new input outside training range, it still falls in some leaf → gets that leaf's constant. The prediction is bounded by training data min/max.

### Q47: How does a tree handle interactions between features?
**Answer:** Naturally! Each path from root to leaf represents an AND of conditions (e.g., `Age<30 AND Income>50k AND Credit=Good`). Trees automatically learn feature interactions without explicit specification.

### Q48: What is the minimal description length (MDL) principle in pruning?
**Answer:** A tree is good if it's small (low description length) AND fits data well. Pruning balances tree complexity (description) against training error (fit). Cost-complexity pruning is an MDL-style approach.

### Q49: How would you implement a tree that handles missing values natively?
**Answer:** Two approaches:
1. **Surrogate splits:** For each split, find backup splits that mimic it. If the primary feature is missing, use surrogates.
2. **Send to both children:** Send samples with missing values to both children with fractional weights.

### Q50: How do trees compare to neural networks?
**Answer:**
| | Trees | Neural Networks |
|---|---|---|
| Interpretability | High | Low |
| Data needed | Less | More |
| Scaling | Linear features | High-dimensional |
| Text/Image | Weak | Strong |
| Extrapolation | No | Yes |
| Training speed | Fast | Slow |
| Production | Use ensembles | Direct |

### Q51: What is gradient boosted trees?
**Answer:** Sequentially train trees, each correcting errors of the previous. The final prediction is the sum of all trees. Examples: XGBoost, LightGBM, CatBoost. Often the best performance on tabular data.

### Q52: How does Random Forest reduce variance?
**Answer:** Train many trees on bootstrap samples with random feature subsets, then average predictions. The randomness decorrelates trees, so averaging reduces variance (central limit theorem).

### Q53: What is the "one standard error" rule in pruning?
**Answer:** Choose the simplest tree (smallest) within 1 standard error of the best CV accuracy. Avoids overfitting to CV noise — picks a more robust, simpler tree.

### Q54: How do you explain a single prediction (local interpretability)?
**Answer:** Use SHAP (SHapley Additive exPlanations) or LIME. SHAP attributes the prediction to features based on game theory. For trees, TreeSHAP is exact and fast.

### Q55: When would you NOT use a decision tree?
**Answer:**
1. **Extrapolation needed** (time series forecasting).
2. **Very smooth relationships** (linear is better).
3. **High-dimensional sparse data** (text — use Naive Bayes or Linear SVM).
4. **Image/audio** (use CNNs/RNNs).
5. **Production with single tree** (use ensemble instead).


# Part 20: Common Mistakes

---

## 20.1 Mistake 1: Not Pruning
**Problem:** Tree grows until all leaves are pure → 100% train accuracy, poor test.
**Fix:** Always set `max_depth`, `min_samples_leaf`, or use `ccp_alpha`.

## 20.2 Mistake 2: Using Single Trees in Production
**Problem:** High variance — small data changes → very different trees.
**Fix:** Use Random Forest or Gradient Boosting.

## 20.3 Mistake 3: Not Setting `random_state`
**Problem:** Trees can differ run-to-run (especially with ties in splits).
**Fix:** Always set `random_state=42`.

## 20.4 Mistake 4: Trusting `predict_proba` as True Probabilities
**Problem:** Tree probabilities are leaf class proportions — not calibrated.
**Fix:** Use `CalibratedClassifierCV` if calibration matters.

## 20.5 Mistake 5: Scaling Features (Unnecessary)
**Problem:** Wasted compute; no benefit.
**Fix:** Trees don't need scaling — skip it.

## 20.6 Mistake 6: Using Trees for Extrapolation
**Problem:** Can't predict beyond training range.
**Fix:** Use linear models or neural networks for forecasting.

## 20.7 Mistake 7: Ignoring Feature Importance Bias
**Problem:** High-cardinality features get inflated importance.
**Fix:** Use permutation importance for unbiased estimates.

## 20.8 Mistake 8: Not Handling Class Imbalance
**Problem:** Tree biases toward majority class.
**Fix:** Use `class_weight='balanced'` or resampling.

## 20.9 Mistake 9: One-Hot Encoding Without Pruning
**Problem:** One-hot creates many features → more overfitting.
**Fix:** Aggressive pruning after one-hot encoding.

## 20.10 Mistake 10: Not Using Cross-Validation
**Problem:** Single train/test split gives unreliable tree evaluation.
**Fix:** Always use k-fold CV for tree evaluation.

| Extrapolating | Use linear/NN |
