### Exception Handling

#### Error
1. A problem /mistake in program that causes it to fail or behave incorrectly.
2. Often cannot be handled
3. e.g. SyntaxError, IndentationError

### Exception
1. A runtime problem that occurs while running program which can be handled.
2. It can be handled

1. try -> lets you test a block of code for errors
2. except -> runs when exception occurs, lets you handle error
3. else -> lets you execute code when there is no error
4. finally -> lets you execute block regardlessly
5. e.g. ZeroDivisonError, ValueError, KeyError, NameError, IndexError

```python
try:
    numerator = int(input("Enter any number"))
    denominator = int(input("Enter any number"))
 
    div = numerator / denominator
except:
    print("Denominator cannot be zero")
else:
    print("There is no error")
    print(f"{numerator} / {denominator} = {div}")
finally:
    print("finally block")
```

```python
l = [1,2,3,4,5]

def get_length(l):
    count = 0
    i = 0
    while True:
        try:
            l[i]
            count += 1
        except:
            break
        else:
            i += 1
    return count

print(f"count = {get_length(l)}")

# print(count)
```

### OOP (Object Oriented Programming)
It is a programming way/method which oraganizes code into classes and objects.
1. Encapsulation
2. Polymorphism
3. Abstraction
4. Inheritance
#### Class
1. It is blueprint or template used to create objects.
2. It defines what attributes and methods the objects will have
#### Object
1. It is real/actual instance of the class
2. It contains actual data and can perform the actions defined in class.
   


```python
class Student:

    # constructor function -> gets executed when instance/object of class is created
    def __init__(self, fname="Mohan", lname="Gorale"):
        # attributes
        self.fname = fname
        self.lname = lname
        pass

    def introduce(self):
        print(f"Hello My name is {self.fname} {self.lname}")
    
    # class method
    @classmethod
    def demo(cls):
        print("This is class method of Student")

s1 = Student("abc", "def")
s1.introduce()
Student.demo()
```

```python
class Calculator:
    def __init__(self):
        pass
    @classmethod
    def Add(cls, a, b):
        return a+b
    
    @classmethod
    def Subtract(cls, a, b):
        return a-b

Calculator.Add(2,3)
Calculator.Subtract(2,3)
```

```python
# Parent Class
class Person:
    def __init__(self, fname, lname):
        self.fname = fname
        self.lname = lname
    
    def get_name(self):
        print(f"First name is {self.fname}")
    
    def introduce(self):
        print(f"Hello My name is {self.fname} {self.lname}")
    

# Child Class
class Student(Person):
    def __init__(self, fname, lname, marks):
        # calling constructor function of parent class
        # Person.__init__(self, fname, lname)
        super().__init__(fname, lname)
        self.marks = marks

    def introduce(self):
        print(f"Hello My name is {self.fname} {self.lname} and I am a Student\n I scored {self.marks} marks")



s1 = Student("Omkar", "Gode", 70)
# s1.get_name()
s1.introduce()
```

### File Handling
It is process of creating, opening, reading, writing, appending, updating and closing files stored on computer.

#### Text Files
Stores data as readable characters
e.g. .txt, .csv, .json, .html, .xml

#### Binary files
Store data in binary format
e.g. images (.jpg, .jpeg, .png, .gif), videos (.mp4), Audio (.mp3), executable (.exe)

#### Why it is important ?

```python
# open(filename, mode)

##### File Modes

# r = read mode -> opens file for reading only, error occurs if file doesn't exist
# w = write mode -> creates new file if it doesn't exist, overwrites if exists
# a = append mode -> adds data at the end of the file, creates new file in it doesn't exist
# r+ = read and write -> read and write both
# w+ = read and write -> read and write both but it creates new file 
# a+ = append and read 
# x = creates specified file, shows error if it exists
```
