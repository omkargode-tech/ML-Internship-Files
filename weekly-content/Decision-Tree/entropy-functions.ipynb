```python
import math

def entropy(counts):
    total = sum(counts)

    if total == 0:
        return 0

    probabilities = [count / total for count in counts]

    return -sum(
        p * math.log2(p)
        for p in probabilities
        if p > 0
    )


def weighted_entropy(groups):
    total_samples = sum(sum(group) for group in groups)

    if total_samples == 0:
        return 0

    return sum(
        (sum(group) / total_samples) * entropy(group)
        for group in groups
    )
```

```python
entropy([5,5])
```

```python
left = [2,1]
right = [3,4]

groups = [left, right]

print(weighted_entropy(groups))
```
