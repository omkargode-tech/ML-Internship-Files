### Numpy
Numerical Python

1. Python library used for working with arrays (Vectors, Matrices)
2. It has function for linear algebra, descriptive statistics.

#### why ?
- lists are slow in python 
- numpy array is 50X faster than list
- array of numpy -> ndarray
- Faster for matrix calculations
- stores elements in continuous memory location

#### Datatypes in Numpy
+ int, float, bool(1bit), string, complex, unicode, object
+ Numeric types have precisions like 32-bit or 64-bit
+ Data can be represented using either Type or Type Code
