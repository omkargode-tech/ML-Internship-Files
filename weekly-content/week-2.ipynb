### Recursion
It is a function calling itself repeatedly to solve a specific problem.
1. recursive case -> when we want to call function again
2. base case -> when we want to stop recursion

```python
# Print below sequence using Recursion
# 5 4 3 2 1

def demo1(n):
    # local scope
    if n < 1:
        # base case
        return
    else:
        # recursive case
        print(n)
        demo1(n-1)


demo1(5)
```

```python
# Sum of First n numbers using Recursion

def get_sum(num):
    if num < 1:
        return 0
    else:
        return num + get_sum(num-1)


print(get_sum(5))
```

### List
1. It is an ordered, mutable/changeable collection of items/elements.
2. It is dynamic -> size is fixed
3. Allows duplicates, can be modified.
4. can have elements of any data type.

```python
# Creating List

l = [1,3.14,"Indala",True]
# l = list((1,2,3,"I",3.3))
```

```python
# Accessing List elements
l[2]

# traversing -> going through all elements of collection
for element in l:
    print(element)
```

```python
# Adding elements to the list

l[2] = "College"

# adds value to the end of the list
l.append("Omkar")

# adds value to the specified index/location
l.insert(1, 10)

# adds another list to currrent list
l.extend([30,40,50])

l
```

```python
# Removing elements from list

# deletes element at index 5 using "del" keyword
del l[5]

# removes element which is passed as argument
l.remove(10)

# removes element at passed index
l.pop(3)

# removes all elements but list remains
l.clear()

# deletes both elements and list
del l
```

```python
# Copy fuction is used to copy list from one variable to other
# Otherwise changes made in list2 will be made in list1

l = [3,4,5,1,2]
l2 = l.copy()

l2.append("Indala")

print(l)
print(l2)
```

### Set
1. It is unordered, mutable collection of unique elements.
2. Automatically removes duplicate values

```python
# Creating Sets
s1 = set((1,2,3))
s2 = {3,4,5}

print(s1)
print(s2)
```

```python
# traversing set
for i in s1:
    print(i)
```

```python
# Adding elements to sets
s1.add(4)
s1.update(s2)
s1
```

```python
# Removing elements from sets
s1.remove(5)
s1.discard(4)
s1
```

```python
# Union of two sets using union function
s1 = set((1,2,3))
s2 = {3,4,5}
s3 = s1.union(s2)
# s1.update(s2)
print(s1)
print(s3)
```

```python
# Intersection of two sets using intersection function
s1 = set((1,2,3))
s2 = {3,4,5}
s3 = s1.intersection(s2)

print(s3)
```

```python
# Difference of two sets using difference function
s1 = set((1,2,3))
s2 = {3,4,5}
s3 = s1.difference(s2)

print(s3)
```

```python
# Find the sum of numbers from 1 to n.
# take input from user - number e.g. 10 

n = int(input("Enter any number: "))
sum = 0

for i in range(1, n+1):
    sum = sum + i
    print(sum)

print(f"Sum is first {n} numbers is {sum}")
```

```python
# Write a function that returns the reverse of a string.  without using ->  s[::-1], s.reverse()

def reverse_string(st):
    li = len(st) - 1
    reverse_string = ""

    while li >= 0:
        reverse_string += st[li]
        li = li - 1

    return reverse_string


reverse_string("abcdefgh")
```

```python
# Leap Year checker
n = int(input("Enter a year: "))
if (n % 4 == 0):
    print(f"It is a leap year")
else:
    print(f"It is not a leap year")
```

```python
# Prime Number checker

n = int(input("Enter any number: "))

def is_prime_number(num):
    flag = True
    i = 2

    if num == 2:
        print(f"{num} is a prime number")
        return

    while i < num:
        if num % i == 0:
            flag = False
            break
        i += 1

    if flag == True:
        print(f"{num} is a prime number")
    else:
        print(f"{num} is a not prime number")

is_prime_number(n)
```

```python
# Palindrome Checker

# racecar -> racecar
# Indala -> aladnI

s = input("Enter any string: ")
# rs = s[::-1]

si = 0
ei = len(s) - 1
is_palindrome = True

while si <= ei:
    if s[si] != s[ei]:
        is_palindrome = False
        break
    si += 1
    ei -= 1

if is_palindrome == True:
    print(f"{s} is palindrome")
else:
    print(f"{s} is not palindrome")
```

```python
# Number Guessing Game

num = int(input("Enter any number from 1 to 100: "))

flag = True
while flag == True:
    guess = int(input("Guess the number: "))
    if guess == num:
        print("You guessed correctly")
        break
    elif guess > num:
        print("Number is too high")
    else:
        print("Number is too low")
```

```python
# Counting input character in string

inp_str = "aa bbb c d g ihj zzzzzz"
inp_chr = input("Enter any character: ")
count = 0

for i in range(0, len(inp_str)):
    if  inp_chr == inp_str[i]:
        count += 1

print(f"{inp_chr} appears {count} times in {inp_str}")
```
