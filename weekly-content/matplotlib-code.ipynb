# Matplotlib
It is visualization library in python.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Line Plot in Matplotlib

x1 = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]

x2 = [1, 2, 3, 4, 5]
y2 = [1, 2, 3, 4, 5]

plt.figure(figsize=(7,4))

plt.plot(x1, y1, marker="o", color="#2917FF", linestyle="--")
plt.plot(x2, y2, marker="^", color="green", linestyle="-")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Heading (Title)")

plt.show()
```

```python
data = {
    "Month": [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ],
    "Sales": [120, 135, 150, 170, 160, 180, 210, 220, 205, 230, 250, 270],
    "Expenses": [80, 90, 95, 100, 98, 110, 120, 125, 118, 130, 140, 150],
    "Profit": [40, 45, 55, 70, 62, 70, 90, 95, 87, 100, 110, 120]
}

df = pd.DataFrame(data)

print(df)
```

Line Plot

```python
# Line plot graph

plt.figure()


plt.plot(df["Month"], df["Sales"], color="orange", label="Sales")
plt.plot(df["Month"], df["Profit"], label="Profit")

# for labeling x axis
plt.xlabel("Month", fontweight="bold", labelpad=15)
# for labeling x axis
plt.ylabel("Sales and Profit", fontweight="bold", labelpad=15)

plt.xticks(fontsize=10, rotation=90)

plt.title("Monthly Profit vs Sales ", fontsize=16, pad=20, family="cursive")

plt.grid()

plt.legend()
# plt.savefig("line-plot.png", dpi=100)

plt.show()
```

```python
study_data = {
    "study_hours": [
        1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 5.5, 6.0,
        6.5, 7.0, 7.5, 8.0, 8.5, 9.0, 9.5, 10.0, 10.5, 11.0
    ],
    "exam_score": [
        42, 48, 50, 56, 58, 63, 66, 69, 72, 75,
        78, 81, 84, 87, 89, 92, 94, 96, 97, 99
    ]
}


plt.figure()

plt.scatter(x=study_data["study_hours"],
            y=study_data["exam_score"],
            s=50,
            color='red', 
            alpha=0.5)

plt.title("Study hours vs Marks", fontweight="bold", pad=15, fontsize=16)

plt.xlabel("Study Hours", labelpad=15)
plt.ylabel("Exam Score", labelpad=15)

plt.show()
```

Bar Chart

```python
x = ["Mango", "Watermelon", "Lychee", "Orange", "Grapes", "Pineapple", "Apple"]
y = [10, 15, 50, 30, 100, 9, 15]
colors = ["gold", "darkgreen", "lightgray", 'orange', 'lightgreen', 'brown', 'red']

plt.figure()

plt.bar(x=x, height=y,  color=colors)

plt.xlabel("Fruits", labelpad=15)
plt.ylabel("Count", labelpad=15)

plt.title("Fruits Distribution", fontweight="bold", pad=15, fontsize=16)

plt.xticks(fontsize=10, rotation=90)

plt.show()
```

Horizontal Bar Graph

```python
x = ["Mango", "Watermelon", "Lychee", "Orange", "Grapes", "Pineapple", "Apple"]
y = [10, 15, 50, 30, 100, 9, 15]
colors = ["gold", "darkgreen", "lightgray", 'orange', 'lightgreen', 'brown', 'red']

plt.figure(figsize=(10,4))

plt.barh(y=x, width=y,  color=colors)

plt.xlabel("Counts", labelpad=15)
plt.ylabel("Fruits", labelpad=15)

plt.title("Fruits Distribution", fontweight="bold", pad=15, fontsize=16)

plt.xticks(fontsize=10, rotation=90)

plt.show()
```

Histogram

```python
marks = np.random.normal(70, 20, 100)

plt.hist(marks,
         edgecolor="white",
         alpha=0.7)

plt.title("Marks Distribution", fontweight="bold", pad=15, fontsize=16)

plt.xlabel("Marks", labelpad=15)
plt.ylabel("Frequency", labelpad=15)

plt.show()
```

Pie Chart

```python
no_students = [97, 85, 77, 40]
stream = ["IT", "CO", "AI", "Mech"]

colors = ['gold', 'coral', 'skyblue', 'lightgreen']
explode = [0.1, 0, 0, 0]

plt.figure()
plt.pie(no_students, 
        labels=stream, 
        colors=colors, 
        explode=explode, 
        wedgeprops={"edgecolor":"white", "linewidth":5, "width": 0.7},
        textprops={"fontweight":"bold"})

plt.title("Total Students", fontweight="bold", pad=15, fontsize=16)

plt.show()
```

Box Plot

```python
marks = [81, 40, 56, 67, 51, 73, 94, 54, 48, 53, 40, 69, 45, 57, 67, 64, 60, 59, 56, 77, 4, 100, 39, 42]

# ss = pd.Series(marks)
# print(ss.describe())

plt.figure()

plt.boxplot(marks,
            patch_artist=True,
            vert=False,
            boxprops={"facecolor":'lightblue'})

plt.xlabel("Marks")
# plt.ylabel("Box plot")
plt.title("Box Plot", fontweight="bold", pad=15, fontsize=16)

plt.show()

```
